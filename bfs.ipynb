```python
def bfs(adjl,s=0):
    q=Queue()
    visited=set()
    result=[]
    q.enqueue(s)
    while not q.isEmpty():
        for j in adjl[q.peek()]:
            if j not in visited:
                q.enqueue(j)
        result.append(q.peek())
        visited.add(q.dequeue())
    return result
```

```python
class Queue:
    def __init__(self):
        self.queue=[]
    def enqueue(self,value):
        self.queue.append(value)
    def dequeue(self):
        if self.isEmpty():
            return "Queue is empty!!"
        return self.queue.pop(0)
    def peek(self):
        if self.isEmpty():
            return "Queue is empty!!"
        return self.queue[0]
    def isEmpty(self):
        return len(self.queue)==0
    def size(self):
        return len(self.queue)
    def display(self):
        return self.queue
```
